```python
# DataFrame 2: Order Information 
# customer_id 6 is not in customers_df

orders_df = pd.DataFrame({'order_id': [101, 102, 103, 104, 105, 106], 
                          'customer_id': [1, 3, 2, 6, 1, 3], 
                          'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Charger', 'Webcam'], 
                          'amount': [1200, 25, 75, 300, 50, 40] }) 


orders_df
```
