```python
from abc import abstractmethod, ABC
```

```python
# class contract
class FileSystemManager(ABC):

    @abstractmethod
    def list_files(self,glob_expression:str):
        pass

    @abstractmethod
    def put_file(self, local_file_path:str, target_file_path:str):
        pass


import shutil


# implementation of the contract 
# inheritance
class LocalFileSystemManager(FileSystemManager):

    def __init__(self, current_working_directory):
        self.current_working_directory = current_working_directory

    
    def list_files(self,glob_expression:str = "*"):
        return glob(os.path.join(self.current_working_directory, glob_expression))


    def put_file(self, name:str, target_directory:str):
        local_file_path = os.path.join(self.current_working_directory, name)
        shutil.copy(local_file_path, target_directory)

fsm = LocalFileSystemManager("/home/aio/training/data/")
fsm.list_files()
```

```python
# implementation of the contract 
# inheritance
from time import sleep

class LocalFileSystemManager(FileSystemManager):

    def __init__(self, current_working_directory):
        self.current_working_directory = current_working_directory

    
    def list_files(self,glob_expression:str = "*"):
        return glob(os.path.join(self.current_working_directory, glob_expression))


    def put_file(self, name:str, target_directory:str):
        local_file_path = os.path.join(self.current_working_directory, name)
        if not os.path.exists(local_file_path):
            raise RuntimeError(f"File is not found: {name}")
        for i in range(3): # retry log
            try:
                shutil.copy(local_file_path, target_directory)
            except Exception as e:
                print(f"Attempt:{i}, Failed to copy the file. " + str(e))
                sleep(1.0)

fsm = LocalFileSystemManager("/home/aio/training/data/")
fsm.put_file("diamonds.csv", "/tmp/demo")
```

```python
# implementation of the contract 
# inheritance
from time import sleep

class FileNotFoundException(Exception): 
    pass

class LocalFileSystemManager(FileSystemManager):

    def __init__(self, current_working_directory):
        self.current_working_directory = current_working_directory

    
    def list_files(self,glob_expression:str = "*"):
        return glob(os.path.join(self.current_working_directory, glob_expression))


    def put_file(self, name:str, target_directory:str):
        local_file_path = os.path.join(self.current_working_directory, name)
        if not os.path.exists(local_file_path):
            raise FileNotFoundException(f"File is not found: {name}")
        max_attempts = 3
        for i in range(max_attempts): # retry log
            try:
                shutil.copy(local_file_path, target_directory)
            except Exception as e:
                if i == max_attempts-1:
                    raise e
                print(f"Attempt:{i}, Failed to copy the file. " + str(e))
                sleep(1.0)

fsm = LocalFileSystemManager("/home/aio/training/data/")
fsm.put_file("diamonds.csv", "/tmp/demo")
```

```python
class Book:
    def __init__(self, name:str, author:str):
        print("Constructor of Book")
        self.name = name 
        self.author = author

    def __repr__(self):
        return f"Book(name={self.name}, author={self.author})"

    def __eq__(self, other):
        assert(isinstance(other, Book))
        return self.name == other.name and self.author == other.author # case sensitive
        
b1 = Book("World is flat", "Tom Friedman") # class instance1
b2 = Book("World is flat", "Tom Friedman") # class instance2

b1 == b2
```

```python
class Book:
    def __init__(self, name:str, author:str):
        print("Constructor of Book")
        self.name = name 
        self.author = author

    def __repr__(self):
        return f"Book(name={self.name}, author={self.author})"

    def __eq__(self, other):
        assert(isinstance(other, Book))
        return self.name.lower() == other.name.lower() and self.author.lower() == other.author.lower() # case insensitive
        
b1 = Book("World is flat", "Tom Friedman") # class instance1
b2 = Book("World is flat", "tom friedman") # class instance2

b1 == b2
```
