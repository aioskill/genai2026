```python
def add_all(*args)->int:
    print(type(args))
    total = 0
    for v in args:
        total += v
    return total

add_all(3, 4, 6, 7, 8)
```

```python
def connect_to_database(**kwargs)->int:
    print(type(kwargs))
    print(kwargs)

connect_to_database(db = "dev", host = "server01", user = "app", password = "pass123")
```

```python
def f(a, b, c, *args, **kwargs)->int:
    print("a", a)
    print("b", b)
    print("c", c)
    print("args", type(args))
    print("args", args)
    print("kwargs", type(kwargs))
    print("kwargs", kwargs)

f(1, 2, 3, 4, 5, 6, db = "dev", host = "server01", user = "app", password = "pass123")
```

```python
ts = time() # return epochs in seconds
ts
```

```python
# decorator 


def timeit(func):
    def wrapper(*args, **kwargs):
        ts = time()
        r = func(*args, **kwargs)
        duration = time()-ts
        print("time taken: ", duration, "seconds")
        return r
    return wrapper
```

```python
@timeit
def my_function():
    print("before sleep")
    sleep(2.0)
    print("after sleep")

my_function()
```
