```python
def iqr(a):
    q1, q3 = np.percentile(a, [25, 75])
    return q3-q1
```

```python
a = [6, 10, 8, 11, 13, 15, 6]
np.mean(a), np.median(a), np.max(a)-np.min(a), iqr(a)
```

```python
a = [6, 10, 8, 11, 13, 15, 6, 100]
np.mean(a), np.median(a), np.max(a)-np.min(a), iqr(a)
```

```python
# median is less impacted by the outliers/extreme values
# hence, median is considered a "robust" statistic
# iqr is robust statistic to measure dispersion .. while range, std, var are not.
```
