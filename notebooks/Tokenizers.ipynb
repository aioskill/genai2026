```python
# View vocabulary

print("Encoding:", enc.name)
print("Vocabulary size:", enc.n_vocab)

# Display first 100 token IDs
start = 1000
for token_id in range(start, start+100):
    token_bytes = enc.decode_single_token_bytes(token_id)

    print(
        token_id,
        repr(token_bytes),
        token_bytes.decode("utf-8", errors="replace")
    )
```

```python
with open("/tmp/vocabulary.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["token_id", "token_text", "token_bytes_hex"])

    for token_id in range(enc.n_vocab):
        try:
            raw = enc.decode_single_token_bytes(token_id)
        except KeyError:
            continue

        writer.writerow([
            token_id,
            raw.decode("utf-8", errors="replace"),
            raw.hex()
        ])
```
