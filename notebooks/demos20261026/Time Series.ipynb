```python
csco = df[df.symbol == 'CSCO'][['adjclose']].asfreq("B")
csco.sort_index(ascending=False)
```

```python
csco = df[df.symbol == 'CSCO'][['adjclose']].asfreq("B")
csco['adjclose_lag1'] = csco['adjclose'].shift()
csco
```

```python
csco = df[df.symbol == 'CSCO'][['adjclose']].asfreq("B")
csco['adjclose_lag1'] = csco['adjclose'].shift(2)
csco
```

```python
csco = df[df.symbol == 'CSCO'][['adjclose']].asfreq("B")
csco['adjclose_diff1'] = csco['adjclose'].diff(1)
csco
```

```python
csco = df[df.symbol == 'CSCO'][['adjclose']].asfreq("B")
csco['adjclose_pct2'] = csco['adjclose'].pct_change(2)
csco
```

```python
csco = df[df.symbol == 'CSCO'][['adjclose']].asfreq("B")
csco['adjclose_pct1'] = csco['adjclose'].pct_change(1)
csco['adjclose_pct2'] = csco['adjclose'].pct_change(2)
csco['adjclose_pct3'] = csco['adjclose'].pct_change(3)
csco['adjclose_pct4'] = csco['adjclose'].pct_change(4)
csco
```

```python
csco = df[df.symbol == 'CSCO'][['adjclose']].asfreq("B")
csco['adjclose_rolling_mean3'] = csco['adjclose'].rolling(3).mean() # moving avg 
csco
```
