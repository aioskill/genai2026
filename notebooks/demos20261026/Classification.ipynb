```python
from sklearn import linear_model, preprocessing, model_selection, metrics, pipeline
```

```python
X = pd.get_dummies(df.drop(columns=["target", "target2", ]))
y = df['target2']

X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size = 0.2, random_state=123)

pipe = pipeline.Pipeline([
    ('scaler', preprocessing.StandardScaler()),
    ('est', linear_model.LogisticRegression(max_iter=1000)),
])

pipe.fit(X_train, y_train)
```

```python
y_train_pred = pipe.predict(X_train)
y_test_pred = pipe.predict(X_test)
```

```python
pd.DataFrame({'y': y_test, 'y_pred': y_test_pred})
```

```python
metrics.confusion_matrix(y_test, y_test_pred)
```

```python
print("accuracy", metrics.accuracy_score(y_test, y_test_pred))
print("precision", metrics.precision_score(y_test, y_test_pred))
print("recall", metrics.recall_score(y_test, y_test_pred))
```

```python
# First column P(y=0), second column is P(y=1)
a = pipe.predict_proba(X_train)
a
```

```python
a.sum(axis=1)
```

```python
y_train_prob = pipe.predict_proba(X_train)[:, 1]
y_test_prob = pipe.predict_proba(X_test)[:, 1]
```

```python
y_test_prob
```

```python
y_test_pred = np.where(y_test_prob>0.5, 1, 0)
print("accuracy", metrics.accuracy_score(y_test, y_test_pred))
print("precision", metrics.precision_score(y_test, y_test_pred))
print("recall", metrics.recall_score(y_test, y_test_pred))
print("confusion\n", metrics.confusion_matrix(y_test, y_test_pred))
```
