```python
import numpy as np

# def find_outliers(a):
#     q1, q2, q3 = np.percentile(a, [25, 50, 75])
#     iqr = q3 - q1
#     upper_whisker = min(a.max(), q3 + 1.5 * iqr)
#     lower_whisker = max(a.min(), q1 - 1.5 * iqr)
#     return (a > upper_whisker) | (a<lower_whisker)

def find_outliers(a):
    return a > a.mean()+ 3*a.std()
```

```python
a = df.outflow_Router
a[a>a.mean()+3 * a.std()]
```

```python
a = df.drop(columns="timestamp").reset_index()
a[find_outliers(a.outflow_Router)]
```

```python
frames = []
for link in df.link.unique():
    a = df[df.link == link]
    outliers = find_outliers(a.outflow_Router)
    print("link", link, sum(outliers))
    a.loc[outliers, 'outflow_anom'] = True
    frames.append(a)
df = pd.concat(frames)
df
```
