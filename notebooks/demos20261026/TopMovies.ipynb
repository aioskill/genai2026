```python
m = movies[movies.title == 'Godfather, The (1972)'].movieId.tolist()[0]
m
```

```python
avg_ratings = ratings.groupby("movieId").rating.agg(["count", "mean"]).query('count>100')
avg_ratings
```

```python
top10 = avg_ratings.sort_values("mean", ascending=False).head(10)
top10
```

```python
movies.head()
```

```python
top10.merge(movies, left_index=True, right_on="movieId")
```

```python
movies['genres'] = movies.genres.str.split("|")
movies
```

```python
movies.explode("genres")
```

```python
movies.explode("genres").genres.value_counts()
```

```python
movies.explode("genres").genres.value_counts()[:3]
```

```python
a = movies.explode("genres").merge(avg_ratings, left_on = "movieId", right_index = True)
a
```

```python
a['rank'] = a.groupby("genres")["mean"].rank(ascending = False)
a.query('rank <= 3').sort_values(["genres", "mean"], ascending = [True, False])
```
