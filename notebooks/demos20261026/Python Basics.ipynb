```python
# Closure: track how much a function call takes
def timeit(f:callable):
    def wrapper(*args, **kwargs):
        start_ts = time()
        out = f(*args, **kwargs)
        print("Time taken: ", (time()-start_ts))
        return out
    return wrapper
```

```python
def check_sql_injection(f):
    def wrapper(sql:str, *args, **kwargs):
        if 'DELETE' in sql.upper():
            raise ValueError('Invalid SQL. Found injection attack')
        return f(sql, *args, **kwargs)
    return wrapper
```
