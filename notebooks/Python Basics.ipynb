```python
# decorator

def timeit(f:callable):
    def wrapper(*args, **kwargs):
        response = f(*args, **kwargs)
        return response
    return wrapper

@timeit
def f():
    sleep(3.0)

f()
```

```python
# decorator
from time import time

def timeit(f:callable):
    def wrapper(*args, **kwargs):
        start_ts = time()
        response = f(*args, **kwargs)
        duration = time() - start_ts
        print("time taken: ", duration, "s")
        return response
    return wrapper

@timeit
def f():
    sleep(3.0)

f()
```

```python
# File system, magic methods 
class File:

    # constructor
    def __init__(self, file_name:str, path:str):
        self.file_name = file_name
        self.path = path

    def __eq__(self, other):
        assert isinstance(other, File)
        print("eq is checked")
        return (self.file_name == other.file_name) and (self.path == other.path)
    
f1 = File('sample.txt', '/tmp/sample.txt') # class instance, object
f2 = File('sample.txt', '/tmp/sample.txt') # class instance, object
f1 == f2
```

```python
from abc import ABC, abstractmethod
from typing import List 
from glob import glob
import shutil

@dataclass
class FileServiceConfig:
    file_system:str = 'Local file sysem'
    root_path:str = '/tmp'

@dataclass
class File:
    file_name:str
    path:str
    
class FileService(ABC):

    def __init__(self, config: FileServiceConfig):
        self.config = config

    @abstractmethod
    def list_files(self, path)-> List[File]:
        pass

    @abstractmethod
    def cp(self, src, dst):
        pass

    @abstractmethod
    def file_size(self, path):
        pass

    @abstractmethod
    def delete_file(self, path):
        pass


# Class inheritance
class LocalFileService(FileService):

    def __init__(self, config: FileServiceConfig):
        self.config = config

    def list_files(self, path)-> List[File]:
        return glob(path)

    def cp(self, src, dst):
        pass

    def file_size(self, path):
        pass

    def delete_file(self, path):
        pass

config = FileServiceConfig()
fs = LocalFileService(config)

fs.list_files("/tmp/")
```

```python
from glob import glob 
import os.path

class FileExplorer:

    def __init__(self, path):
        self.path = path
        self.files = glob(os.path.join(path, "*"))

    def __iter__(self):
        for f in self.files:
            yield f
        
for r in FileExplorer("/home/aio/training/data/bluebikes/"):
    print(r)
```
