```python
data = {
    'Category': ['Electronics', 'Electronics', 'Clothing', 'Electronics', 'Clothing', 'Electronics', 'Food', 'Food', 'Clothing', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'South', 'North', 'West', 'East', 'North', 'South'],
    'Sales': [100, 150, 200, 120, 90, 180, 50, 70, 110, 130],
    'Quantity': [10, 15, 20, 12, 9, 18, 5, 7, 11, 13],
    'Rating': [4.5, 3.8, 4.2, 4.0, 3.5, 4.7, 3.0, 4.1, np.nan, 3.9] # Added a NaN
}
df = pd.DataFrame(data)
df
```

```python
df['Avg_Category_Sales'] = df.groupby('Category')['Sales'].transform('mean')
```

```python
df
```

```python
df['Cumulative_Sales_Region'] = df.groupby('Region')['Sales'].cumsum()
df
```

```python
df['Sales_Rank_in_Category'] = df.groupby('Category')['Sales'].rank(method='min', ascending=False)
df
```

```python
a = duckdb.sql('''select 
* ,
rank() over(partition by Region order by Sales desc) Sales_Rank_in_Category
from df
''')
a
```
